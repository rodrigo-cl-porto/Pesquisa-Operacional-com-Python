# ÁRVORE GERADORA MÍNIMA

O Problema da Árvore Geradora Mínima (_Minimal Spanning Tree_), abreviado como AGM ou MST, consiste em encontrar dentro de um grafo $G(v,a)$ um subgrafo $A(v,a) \in G(v,a)$ que possua o menor custo, em que $A$ é um grafo do tipo árvore que contém todos os vértices $v \in G$ mas não necessariamente contém todos as arestas $a \in G$.

Encontrar a AGM pode ser interessante nos casos em que desejamos diminuir o custo de um sistema que pode ser representado por um grafo. Um bom exemplo seria o de encontrar dentro de uma rede de instalações elétricas o subconjunto de fios que necessitará menos recursos para a sua instalação inicial. Ou então selecionar dentre as várias rotas de aviação que interligam um conjunto de aeroportos as rotas que interligam todos os aeroportos que possuem o menor custo de transporte. Os algoritmos mais utilizados para encontrar a AGM são os de [Kruskal](https://pt.wikipedia.org/wiki/Algoritmo_de_Kruskal#:~:text=O%20algoritmo%20de%20Kruskal%20%C3%A9,arestas%20da%20%C3%A1rvore%2C%20%C3%A9%20minimizado.) e de [Prim](https://pt.wikipedia.org/wiki/Algoritmo_de_Prim).

O modelo de otimização linear deste problema é dado como se segue:

$$
\text{Min} \quad \sum_{(v,w) \in E}{c_{vw}x_{vw}}
$$

$$
\begin{array}{rll}
    \text{S.t.:} & & \\
    & \sum_{w \in V \backslash \{r\}}{x_{rw}} \ge 1 & \\
    & \sum_{v \in V}{x_{vr}} = 0 & \\
    & \sum_{v \in V}{x_{vw}} = 1 & \quad \forall w \in V \backslash \{r\} \\
    & u_r = 1 & \\
    & u_w \ge u_v + 1 - n(1 - x_{vw}) + (n-2)x_{wv} & \quad \forall (v,w) \in E \\
    & 1 \le u_v \le n & \quad \forall v \in V \\
    & x_{vw} \in \{0,1\} & \quad \forall (v,w) \in E \\
    & u_v \in \mathbb{N} & \quad \forall v \in V \\
\end{array}
$$

## Resolvendo o problema

````python
import matplotlib.pyplot as plt
import networkx as nx
import pyomo.environ as pyo
from pyomo.core.util import sum_product
````

````python
class MinimumSpanningTree(pyo.ConcreteModel):
    def __init__(self, G: nx.Graph) -> None:
        super().__init__()
        self._graph = G.to_directed()
        self.__set_composite_sets()
        self.__set_parameters()
        self.__set_decision_variables()
        self.__set_objective_function()
        self.__set_constraints()

    def __set_composite_sets(self) -> None:
        self.V = pyo.Set(initialize=list(self._graph.nodes))
        self.r = list(self._graph.nodes)[1]
        self.E = pyo.Set(initialize=list(self._graph.edges), dimen=2)

    def __set_parameters(self) -> None:
        self.n = pyo.Param(
            initialize=self._graph.number_of_nodes(),
            domain=pyo.PositiveIntegers
        )
        self.c = pyo.Param(
            self.E,
            initialize={
                (edge[0], edge[1]): edge[2]
                for edge in self._graph.edges(data="distance")
            },
            domain=pyo.Reals
        )

    def __set_decision_variables(self) -> None:
        self.x = pyo.Var(self.E, domain=pyo.Binary, initialize=0)
        self.u = pyo.Var(self.V, domain=pyo.PositiveIntegers, initialize=1, bounds=(1, self.n))

    def __set_objective_function(self) -> None:
        self.obj = pyo.Objective(sense=pyo.minimize, expr=sum_product(self.c, self.x, index=self.E))

    def __set_constraints(self) -> None:
        self.__set_node_constraints()
        self.__set_branching_levels_constraints()

    def __set_node_constraints(self) -> None:
        root_child_edges = [(v, w) for (v, w) in self.E if v == self.r]
        self.root_must_have_child_constraint = pyo.Constraint(
            expr=pyo.quicksum(self.x[v, w] for (v, w) in root_child_edges) >= 1 
                if root_child_edges else pyo.Constraint.Infeasible
        )

        incoming_edges = {v: [(i, j) for (i, j) in self.E if j == v] for v in self.V}
        self.node_parent_count_constraints = pyo.Constraint(
            self.V, rule=lambda model, v:
                pyo.quicksum(model.x[i, j] for (i, j) in incoming_edges[v]) == (0 if v == model.r else 1)
                if incoming_edges[v] else (pyo.Constraint.Skip if v == model.r else pyo.Constraint.Infeasible)
        )

    def __set_branching_levels_constraints(self) -> None:
        self.root_level_constraint = pyo.Constraint(expr=self.u[self.r] == 1)
        self.mtz_subtour_elimination_constraints = pyo.Constraint(
            self.E, rule=lambda model, v, w:
                model.u[w] >= model.u[v] + 1 - model.n * (1 - model.x[v,w]) + (model.n - 2) * model.x[w,v]
                if w != self.r else pyo.Constraint.Skip
        )

    def solve(self) -> nx.Graph:
        result = pyo.SolverFactory("highs").solve(self)
        is_solved: bool = (
            result.Solver.status == pyo.SolverStatus.ok and ( # Checks if solver is finished with normal termination and if...
                result.Solver.termination_condition == pyo.TerminationCondition.optimal # ended with an optimal solution...
                or result.Solver.termination_condition == pyo.TerminationCondition.feasible # or with a feasible one.
            )
        )
        if is_solved:
            edges = [edge for edge in self.E if pyo.value(self.x[edge]) > 0.5]
            return G.edge_subgraph(edges).to_undirected()

        msg = f"No feasible solution was found!\n\n{result.Solver}"
        raise RuntimeError(msg)
````

## Desenhando os grafos

Para desenhar um grafo, antes de tudo, é necessário criar e guardar as posições dos nós em uma variável. É possível fazer isso manualmente (o que é bem trabalhoso!) ou usar uma função do tipo `_layout()`, já disponível no NetworkX, para fazer isso automaticamente por nós (Ainda bem!). Para isso, usou-se a função [`kamada_kawai_layout()`](https://networkx.org/documentation/stable/reference/generated/networkx.drawing.layout.kamada_kawai_layout.html) e, logo depois, utilizou-se [`draw_networkx()`](https://networkx.org/documentation/stable/reference/generated/networkx.drawing.nx_pylab.draw_networkx.html#networkx.drawing.nx_pylab.draw_networkx) para desenhar os grafos.

````python
# Define o layout dos pontos para exibição
pos = nx.kamada_kawai_layout(G)
nx.draw_networkx(
    G,                       # Grafo original
    ax= plt.axes((0,0,1,1)), # Define o eixo de exibição
    pos= pos,                # Define o layout dos nós         
    with_labels= True,       # Os nós terão rótulos para exibir os seus nomes
    node_color= 'darkblue',  # Define a cor dos nós como azul escuro
    node_size= 300,          # Define o tamanho dos nós
    node_shape= 's',         # Define o formato dos nós como sendo quadrados (square)
    font_weight= 'bold',     # Coloca negrito no texto dos nomes dos nós
    font_color= 'white'      # Define a cor do texto dos nomes como branco
)
plt.title('Grafo original')

# Desenha a AGM
nx.draw_networkx(
    MST,                     # Árvore Geradora Mínima
    ax= plt.axes((1,0,1,1)), # Define o eixo de exibição
    pos= pos,                # Define o layout dos nós         
    with_labels= True,       # Os nós terão rótulos para exibir os seus nomes
    node_color= 'red',       # Define a cor dos nós como vermelho
    edge_color= 'darkred',   # Define a cor das arestas como vermelho escuro
    node_size= 300,          # Define o tamanho dos nós
    node_shape= 's',         # Define o formato dos nós como quadrado
    font_weight= 'bold',     # Coloca negrito no texto dos nomes dos nós
    font_color= 'white'      # Define a cor do texto dos nomes como branco
)
plt.title("Árvore Geradora Mínima")
````
